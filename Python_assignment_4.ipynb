### 1. Write a Python Program to Find the Factorial of a Number?

```python
num = int(input("Enter a number to find factorial:"))
factorial = 1
if num < 0:
    print("Can't calculate factorial of negative number")
elif num == 0:
    print("Factorial of 0 is 1")
else:
    for i in range(1, num+1):
        factorial = factorial*i
print("Factorial of {} is {}".format(num, factorial))
    

          
```

### 4. Write a Python Program to Check Armstrong Number?

Armstrong number is a number that is equal to the sum of power of its digits
- 153 = length = 3
   - 1*1*1 = 1
   - 5*5*5 = 125
   - 3*3*3 = 27
   sum = 1 + 125 + 27 = 153
   num = sum

```python
num = int(input("Enter a number:"))
power = len(str(num))
temp = num
sum = 0
while temp > 0:
    digit = temp%10
    sum = sum + digit ** power
    temp = temp//10
if num == sum :
    print("Number is armstrong number")
else:
    print("Not an armstrong number")

          
```

### 5. Write a Python Program to Find Armstrong Number in an Interval?

```python
## fining for a range of numbers
low = int(input("Lower limit : "))
up = int(input("Upper limit : "))


for num in range(low,up+1):
    power = len(str(num))
    temp = num
    sum = 0
    while temp > 0:
        digit = temp % 10
        sum = sum + digit ** power
        temp = temp//10
    if num == sum :
        print(num)
```
